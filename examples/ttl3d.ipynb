# ttl3d in a notebook

One call turns RDF files, in-memory `rdflib` graphs or tables into an interactive 3D/2D page, shown right here. Drag to rotate, scroll to zoom, click the legend to filter, click a node for its card, switch views at the top.

```python
%pip install "ttl3d>=0.9.0" -q
```

```python
from pathlib import Path

import rdflib
import ttl3d

RAW = "https://raw.githubusercontent.com/soheilabadifard/TTL_to_3D/main/examples/"


def source(name):
    local = Path("examples") / name          # a clone of the repository
    return str(local) if local.exists() else RAW + name


planets = rdflib.Graph().parse(source("solar-system.ttl"), format="turtle")
missions = rdflib.Graph().parse(source("solar-system-missions.ttl"), format="turtle")
ttl3d.show([("solar-system", planets), ("missions", missions)])
```

The options are the command line's: `view`, `color_by`, `layout`, `labels`, `title`, `lang`, `type_links`, `attribute_preds`, `fmt`, `focus`, `hops`, `schema`, `columns`, `base`, and `height` for `show`. Each output stores the whole page (about 1.9 MB) in the notebook file.

```python
ttl3d.show([("solar-system", planets), ("missions", missions)], view="2d", color_by="type", height=500)
```

A pandas DataFrame is a table: one with an `id` column holds nodes, one with `source` and `target` columns edges, and the other columns are properties (an edge's show on the card under its relation). `dtype=str` keeps the cells as written; without it pandas reads a column with blanks as floats, and 1969 shows as 1969.0.

```python
import pandas as pd

nodes = pd.read_csv(source("missions.csv"), dtype=str)
edges = pd.read_csv(source("mission-links.csv"), dtype=str)
ttl3d.show([("missions", nodes), ("mission-links", edges)], color_by="type", height=500)
```

`ttl3d.write(sources, "page.html", ...)` writes the same page to a file, and `ttl3d.to_html(...)` returns it as a string.
